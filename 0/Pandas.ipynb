```python
# ایمپورت کتابخانه‌ها (قراردادِ مرسوم: pd ، np ، plt)
import pandas as pd              # کار با جدول‌ها
import numpy as np               # محاسبات عددی
import matplotlib.pyplot as plt  # رسم نمودار

# نمایش نسخه‌ها
print("pandas     :", pd.__version__)
print("numpy      :", np.__version__)
import matplotlib
print("matplotlib :", matplotlib.__version__)
```

<div dir="rtl" style="text-align:right;font-family:Vazirmatn,Tahoma,'Segoe UI',sans-serif;line-height:2;font-size:15px;"><h4 style="color:#2563eb;margin:14px 0 0 0;">🔹 مثال 1: ساخت اولین DataFrame از یک دیکشنری 🧾</h4></div>

```python
# کلیدهای دیکشنری ← نام ستون‌ها ، لیست‌ها ← مقدارهای هر ستون
data = {
    "name":  ["Ali", "Sara", "Reza"],
    "age":   [22, 21, 23],
    "score": [88, 2, 75],
}

df_demo = pd.DataFrame(data)   # ساخت DataFrame
df_demo
```

<div dir="rtl" style="text-align:right;font-family:Vazirmatn,Tahoma,'Segoe UI',sans-serif;line-height:2;font-size:15px;"><h4 style="color:#2563eb;margin:14px 0 0 0;">🔹 مثال 2: numpy و مقدار گمشده (NaN) 🕳️</h4></div>

```python
number = [10, 20, 30, 40]
sum = 0
for i in number:
    sum +=i
avg = sum/len(number)
print(avg)
```

```python
# np.nan یعنی «مقدار نامشخص / گمشده» ؛ pandas هم برای خانه‌های خالی از همین استفاده می‌کند
arr = np.array([10, 20, 30, 40])
print("میانگین آرایه:", arr.mean())

x = np.nan
print("نوع np.nan :", type(x))        # float است!
print("آیا x گمشده است؟", np.isnan(x))
```

<div dir="rtl" style="text-align:right;font-family:Vazirmatn,Tahoma,'Segoe UI',sans-serif;line-height:2;font-size:15px;"><h4 style="color:#2563eb;margin:14px 0 0 0;">🔹 مثال 3: اولین نمودار با matplotlib 📊</h4></div>

```python
# رسم نمودار میله‌ای نمره‌ی دانشجوها
plt.figure(figsize=(6, 3.5))
plt.bar(df_demo["name"], df_demo["score"], color=["#7c2aed", "#2563eb", "#0d9488"])
plt.title("Student Scores")   # توجه: برای ساده ماندن، برچسب نمودارها را انگلیسی می‌نویسیم
plt.ylabel("score")
plt.xlabel("name")
plt.show()
```

<div style="background:#f5f3ff;border:2px solid #7c3aed;border-top:none;border-radius:0 0 14px 14px;padding:14px 22px;color:#1f2937;">
<b style="color:#7c3aed;">🎯 چه کار می‌کند؟</b><br><b>چندک</b> (کوانتایل) را می‌دهد: مقداری که درصد مشخصی از داده‌ها کمتر از آن هستند. عدد بین ۰ و ۱ می‌گیرد: <code>0.5</code> = میانه ، <code>0.25</code> = چارک اول ، <code>0.75</code> = چارک سوم.<br>
<b style="color:#7c3aed;">✍️ ساختار:</b> <code dir="ltr" style="background:#fff;border:1px solid #7c3aed66;padding:2px 10px;border-radius:6px;color:#111;">df["ستون"].quantile(q)</code><br>
<b style="color:#7c3aed;">💡 کجا به کار می‌آید؟</b> تشخیص داده‌ی پرت، رتبه‌بندی، «۱۰٪ بالای کلاس کیست؟»
</div></div>

<div dir="rtl" style="text-align:right;font-family:Vazirmatn,Tahoma,'Segoe UI',sans-serif;line-height:2;font-size:15px;"><h4 style="color:#7c3aed;margin:14px 0 0 0;">🔹 مثال 1: میانه به‌عنوان چندک ۰٫۵</h4></div>

```python
print(df["gpa"].quantile(0.5), "=", df["gpa"].median())
```

<div dir="rtl" style="text-align:right;font-family:Vazirmatn,Tahoma,'Segoe UI',sans-serif;line-height:2;font-size:15px;"><h4 style="color:#7c3aed;margin:14px 0 0 0;">🔹 مثال 2: چند چندک با هم</h4></div>

```python
df["gpa"].quantile([0.25, 0.5, 0.75])
```

<div dir="rtl" style="text-align:right;font-family:Vazirmatn,Tahoma,'Segoe UI',sans-serif;line-height:2;font-size:15px;"><h4 style="color:#7c3aed;margin:14px 0 0 0;">🔹 مثال 3: دانشجویان ۱۰٪ برتر 🏆</h4></div>

```python
# حد معدل برای ۱۰٪ بالا
top10_limit = df["gpa"].quantile(0.90)
print("حد معدل ۱۰٪ برتر:", round(top10_limit, 2))

df[df["gpa"] >= top10_limit][["name", "gpa"]]
```

<div dir="rtl" style="text-align:right;font-family:Vazirmatn,Tahoma,'Segoe UI',sans-serif;line-height:2;font-size:15px;"><h4 style="color:#7c3aed;margin:14px 0 0 0;">🔹 مثال 4: تشخیص داده‌ی پرت با روش IQR 🚨</h4></div>

```python
Q1 = df["income"].quantile(0.25)
Q3 = df["income"].quantile(0.75)
IQR = Q3 - Q1                       # فاصله‌ی بین چارک اول و سوم

low  = Q1 - 1.5 * IQR               # حد پایین
high = Q3 + 1.5 * IQR               # حد بالا
print(f"بازه‌ی نرمال: {low:.1f} تا {high:.1f}")

# هر کس بیرون از بازه باشد «پرت» است
outliers = df[(df["income"] < low) | (df["income"] > high)]
print("تعداد پرت‌ها:", len(outliers))   # ۰ یعنی داده‌ی پرت نداریم (خبر خوب!)
outliers
```

```python
plt.figure(figsize=(5, 3.5))
plt.boxplot(df["gpa"].dropna())     # dropna: حذف NaN فقط برای رسم
plt.title("GPA box plot")
plt.show()
```

<div dir="rtl" style="text-align:right;font-family:Vazirmatn,Tahoma,'Segoe UI',sans-serif;line-height:2;font-size:15px;">

</div>
<div style="background:#fdf2f8;border:2px solid #db2777;border-top:none;border-radius:0 0 14px 14px;padding:14px 22px;color:#1f2937;">
<b style="color:#db2777;">🎯 چه کار می‌کند؟</b><br>یک <b>نسخه‌ی مستقل</b> از جدول می‌سازد. بدون copy، دو متغیر فقط «دو اسم برای یک جدول» هستند و تغییر یکی، روی دیگری هم اثر می‌گذارد!<br>
<b style="color:#db2777;">✍️ ساختار:</b> <code dir="ltr" style="background:#fff;border:1px solid #db277766;padding:2px 10px;border-radius:6px;color:#111;">df_new = df.copy()</code><br>
<b style="color:#db2777;">💡 کجا به کار می‌آید؟</b> قبل از هر تغییر مهم؛ تا داده‌ی اصلی سالم بماند
</div></div>

<div dir="rtl" style="text-align:right;font-family:Vazirmatn,Tahoma,'Segoe UI',sans-serif;line-height:2;font-size:15px;"><h4 style="color:#db2777;margin:14px 0 0 0;">🔹 مثال 1: دردسر بدون copy 😱</h4></div>

```python
a = pd.DataFrame({"x": [1, 2, 3]})
b = a                      # ❌ کپی نیست! فقط یک اسم دوم برای همان جدول

b.loc[0, "x"] = 999        # فقط b را عوض کردیم...
print(a)                   # ...ولی a هم عوض شد!
```

<div dir="rtl" style="text-align:right;font-family:Vazirmatn,Tahoma,'Segoe UI',sans-serif;line-height:2;font-size:15px;"><h4 style="color:#db2777;margin:14px 0 0 0;">🔹 مثال 2: راه درست با copy ✅</h4></div>

```python
a = pd.DataFrame({"x": [1, 2, 3]})
b = a.copy()               # ✅ کپی مستقل

b.loc[0, "x"] = 999
print("a:", a["x"].tolist())    # دست‌نخورده
print("b:", b["x"].tolist())
```

```python
# ۱) خواندن
data = pd.read_csv("students.csv")

# ۲) شناخت
print("ابعاد:", data.shape)
print("ستون‌ها:", data.columns.tolist())
print("خالی‌ها:\n", data.isna().sum(), "\n")

# ۳) نسخه‌ی کاری
work = data.copy()

# ۴) تمیزکاری
work = work.drop_duplicates()
work["age"]    = work["age"].fillna(work["age"].median())
work["gpa"]    = work["gpa"].fillna(work["gpa"].mean().round(2))
work["score"]  = work["score"].fillna(work["score"].median())
work["income"] = work["income"].fillna(work["income"].median())
work = work.reset_index(drop=True)

# ۵) تحلیل
print("🔸 تعداد شهرها:", work["city"].nunique())
print("🔸 گرایش پرتکرار:", work["major"].mode()[0])
print("🔸 چارک‌های معدل:\n", work["gpa"].quantile([0.25, 0.5, 0.75]).round(2).to_string(), "\n")

# ۶) سه دانشجوی برتر
best = work.sort_values("gpa", ascending=False).head(3)
print("🏆 سه دانشجوی برتر:")
print(best[["name", "major", "gpa"]].to_string(index=False))

# ۷) ذخیره
work.to_csv("final_students.csv", index=False)
print("\n💾 فایل نهایی ذخیره شد:", work.shape)
```

```python
# ۸) نمودار نهایی: میانگین معدل هر گرایش
work.groupby("major")["gpa"].mean().sort_values().plot(
    kind="bar", color=["#2563eb", "#7c3aed", "#059669", "#ea580c"], figsize=(6, 3.5))
plt.title("Average GPA by major")
plt.ylabel("GPA")
plt.xticks(rotation=0)
plt.show()
```
