# Multiple Products

```python
import numpy as np
```

### Inner Product

> <a|b>

```python
def inner_product(a, b):
    return np.vdot(a, b)
```

### Outer Product

> |a><b|

```python
def outer_product(a, b):
    return np.outer(np.asarray(a).reshape(-1), np.asarray(b).reshape(-1).conj())
```

### Matrix Multiply: Zigma Aik.Bkj

> Σ Aik.Bkj

```python
def matrix_product(a, b):
    return np.matmul(a, b)
```

### Tensor Product

> |a> ⊗ |b>

```python
def tensor_product(a, b):
    return np.kron(a, b)
```

### Input Samples

```python
A = np.array([
    [-9, 3j],
    [-1j, 7]
], dtype=complex)

B = np.array([
    [5j, 6j],
    [20, 8]
], dtype=complex)
```

### Example Usages

```python
print("Inner Product:")
print(inner_product(A[0], B[0]))

print("\nOuter Product:")
print(outer_product(A[:, 0], B[:, 0]))

print("\nMatrix Multiplication:")
print(matrix_product(A, B))

print("\nTensor Product:")
print(tensor_product(A, B))
```

# Pauli Matrices

### Pauli-X

> X = σx = 

\begin{bmatrix}
0&1\\
1&0
\end{bmatrix}

> (Quantum Not) (BitFlip)
جابه‌جایی دو حالت پایه کیوبیت


`Χ|0⟩ = |1⟩`
`X|1⟩ = |0⟩`

```python
X = np.array([
    [0, 1],
    [1, 0]
], dtype=complex)
```

### Pauli-Y

> Y = σy =

\begin{bmatrix}
0&-i\\
i&0
\end{bmatrix}

> علاوه بر جابه‌جایی حالت‌ها، فاز مختلط نیز ایجاد می‌کند

`Y|0⟩ = i|1⟩`
`Y|1⟩ = -i|0⟩`

```python
Y = np.array([
    [0, -1j],
    [1j, 0]
], dtype=complex)
```

### Pauli-Z

> Z = σz =

\begin{bmatrix}
1&0\\
0&-1
\end{bmatrix}

> باعث تغییر فاز حالت دوم می‌شود

`Z|0⟩ = |0⟩`
`Z|0⟩ = -|1⟩`

```python
Z = np.array([
    [1, 0],
    [0, -1]
], dtype=complex)
```

### Identity Matrix

```python
I = np.eye(2, dtype=complex)
```

### Output

```python
print("Pauli-X:")
print(X)

print("\nPauli-Y:")
print(Y)

print("\nPauli-Z:")
print(Z)
```

### Hermitian Test

```python
def is_hermitian(matrix):
    return np.allclose(matrix, matrix.conj().T)


print(is_hermitian(X))
print(is_hermitian(Y))
print(is_hermitian(Z))
```

### Unitary Test

```python
def is_unitary(matrix):
    identity = np.eye(matrix.shape[0])

    return np.allclose(matrix.conj().T @ matrix, identity)


print(is_unitary(X))
print(is_unitary(Y))
print(is_unitary(Z))
```

### Trace Test

```python
print(np.trace(X))
print(np.trace(Y))
print(np.trace(Z))
```

### Square Test

```python
print("X²:")
print(X @ X)

print("Y²:")
print(Y @ Y)

print("Z²:")
print(Z @ Z)
```

### Non-Commutativity Test

```python
print("XY:")
print(X @ Y)

print("YX:")
print(Y @ X)

print("Commutator [X,Y]:")
print(X @ Y - Y @ X)
```
